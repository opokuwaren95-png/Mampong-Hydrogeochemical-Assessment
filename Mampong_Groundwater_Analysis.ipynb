```python
import pandas as pd

df = pd.read_csv("Mampong_Master_Dataset.csv")
print(df.columns.tolist())
print(df.head())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.bar(df['Sample_ID'], df['pH'])
plt.axhline(6.5, linestyle='--')
plt.axhline(8.5, linestyle='--')
plt.title("pH of Mampong Groundwater")
plt.xlabel("Sample ID")
plt.ylabel("pH")
plt.savefig("pH_chart.png")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.bar(df['Sample_ID'], df['TDS_mg_L'])
plt.axhline(500, linestyle='--', color='red', label='WHO limit 500 mg/L')
plt.xlabel("Sample ID")
plt.ylabel("TDS mg/L")
plt.title("TDS of Mampong Groundwater")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.savefig("TDS_chart.png", dpi=300)
plt.show()
```

```python
import matplotlib.pyplot as plt

# EC means Electrical Conductivity - shows salt content
plt.figure(figsize=(11,5))
plt.bar(df['Sample_ID'], df['EC_uS_cm'])
plt.axhline(y=1500, color='r', linestyle='--', label='WHO limit 1500 uS/cm')
plt.title("EC of Mampong Groundwater")
plt.xlabel("Sample ID")
plt.ylabel("EC uS/cm")
plt.ylim(0, 1600)  # This fixes the cropping
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.savefig("EC_chart.png", dpi=300)
plt.show()
```

```python
import matplotlib.pyplot as plt

# Turbidity shows how cloudy water is - high means particles
plt.figure(figsize=(11,5))
plt.bar(df['Sample_ID'], df['Turbidity_NTU'])
plt.axhline(y=5, color='r', linestyle='--', label='WHO limit 5 NTU')
plt.title("Turbidity of Mampong Groundwater")
plt.xlabel("Sample ID")
plt.ylabel("Turbidity NTU")
plt.ylim(0, 12)
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()
plt.savefig("Turbidity_chart.png", dpi=300)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Scatter: TDS vs EC - they should correlate (more salt = higher conductivity)
plt.figure(figsize=(7,5))

plt.scatter(df['EC_uS_cm'], df['TDS_mg_L'], s=80)

# Add trendline for thesis
z = np.polyfit(df['EC_uS_cm'], df['TDS_mg_L'], 1)
p = np.poly1d(z)
plt.plot(df['EC_uS_cm'], p(df['EC_uS_cm']), linestyle='--', color='red')

# Label each point with Sample ID
for i, txt in enumerate(df['Sample_ID']):
    plt.annotate(txt, (df['EC_uS_cm'][i], df['TDS_mg_L'][i]), fontsize=8, xytext=(5,5), textcoords='offset points')

plt.xlabel("EC (uS/cm)")
plt.ylabel("TDS (mg/L)")
plt.title("TDS vs EC - Mampong Groundwater")
plt.tight_layout()
plt.savefig("TDS_vs_EC_scatter.png", dpi=300)
plt.show()

# Calculate correlation for your results section
corr = df['EC_uS_cm'].corr(df['TDS_mg_L'])
print(f"Correlation EC vs TDS: {corr:.3f}")
```

```python
import matplotlib.pyplot as plt

# Histogram shows how TDS values are spread across samples
plt.figure(figsize=(7,5))
plt.hist(df['TDS_mg_L'], bins=6, edgecolor='black')

plt.xlabel("TDS mg/L")
plt.ylabel("Number of samples")
plt.title("TDS Distribution - Mampong (n=11)")
plt.axvline(df['TDS_mg_L'].mean(), color='red', linestyle='--', label=f"Mean = {df['TDS_mg_L'].mean():.1f} mg/L")
plt.legend()
plt.tight_layout()
plt.savefig("TDS_histogram.png", dpi=300)
plt.show()

print(f"Mean TDS: {df['TDS_mg_L'].mean():.1f} mg/L")
print(f"Min: {df['TDS_mg_L'].min()} | Max: {df['TDS_mg_L'].max()}")
```
