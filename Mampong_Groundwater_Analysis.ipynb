```python
import pandas as pd

df = pd.read_csv("Mampong_Master_Dataset.csv")
print(df.columns.tolist())
print(df.head())
```

```python
import matplotlib.pyplot as plt

plt.bar(df['Sample_ID'], df['pH'])
plt.axhline(6.5, linestyle='--')
plt.axhline(8.5, linestyle='--')
plt.title("pH of Mampong Groundwater")
plt.xlabel("Sample ID")
plt.ylabel("pH")
plt.savefig("pH_chart.png")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.bar(df['Sample_ID'], df['TDS_mg_L'])
plt.axhline(y=500, color='r', linestyle='--')
plt.title("TDS Chart")
plt.xlabel("Sample ID")
plt.ylabel("TDS mg/L")
plt.savefig("TDS_chart.png")
plt.show()
```

```python
# EC means Electrical Conductivity, shows how much salt in water
plt.bar(df['Sample_ID'], df['EC_uS_cm'])  # Bar chart of EC
plt.axhline(y=1500, color='r', linestyle='--')  # WHO limit for EC is 1500
plt.title("EC Chart")
plt.xlabel("Sample ID")
plt.ylabel("EC uS/cm")
plt.savefig("EC_chart.png")
plt.show()
```

```python
# Turbidity shows how cloudy water is
plt.bar(df['Sample_ID'], df['Turbidity_NTU'])
plt.axhline(y=5, color='r', linestyle='--')  # WHO limit 5 NTU
plt.title("Turbidity Chart")
plt.xlabel("Sample ID")
plt.ylabel("Turbidity NTU")
plt.savefig("Turbidity_chart.png")
plt.show()
```

```python
# Scatter plot to show relation between TDS and EC
plt.scatter(df['TDS_mg_L'], df['EC_uS_cm'])
plt.xlabel("TDS mg/L")
plt.ylabel("EC uS/cm")
plt.title("TDS vs EC Relationship")
# Add labels for each point
for i in range(len(df)):
    plt.text(df['TDS_mg_L'][i], df['EC_uS_cm'][i], df['Sample_ID'][i])
plt.savefig("TDS_vs_EC_scatter.png")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df['TDS_mg_L'], bins=5, color='skyblue', edgecolor='black')
plt.title('Histogram of TDS in Mampong Groundwater')
plt.xlabel('TDS (mg/L)')
plt.ylabel('Number of Samples')
plt.grid(axis='y', alpha=0.3)
plt.savefig('TDS_histogram.png')
plt.show()
```
